# Medical Insurance Cost Prediction Project

## Project Overview

This notebook aims to build and evaluate machine learning models to predict medical insurance charges based on various personal attributes. We will perform data loading, preprocessing, feature engineering, and then train and assess several regression and classification models, including Linear Regression, Logistic Regression, Decision Tree Classifier, Random Forest Classifier, and KMeans Clustering.

## 1. Setup and Imports

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.cluster import KMeans
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, classification_report
```

## 2. Data Loading and Initial Exploration

```python
from google.colab import drive
drive.mount('content_drive')
```

```python
df=pd.read_csv("content_drive/MyDrive/medical_insurance.csv")
```

## 3. Data Preprocessing and Feature Engineering

This section handles categorical feature encoding and numerical type conversions to prepare the data for machine learning models. We will use `LabelEncoder` for categorical features and convert `bmi` and `charges` to integer types.

```python
# Initialize LabelEncoder
coder = LabelEncoder()

# Apply Label Encoding to categorical features
df["sex"] = coder.fit_transform(df["sex"].astype(str))
df["smoker"] = coder.fit_transform(df["smoker"].astype(str))
df["region"] = coder.fit_transform(df["region"].astype(str))

# Convert numerical columns to integer type
df["bmi"] = df["bmi"].astype(int)
df["charges"] = df["charges"].astype(int)

print("Data after preprocessing:")
print(df.head())
```

```python
# Define features (x) and target (y)
x = df.drop("charges", axis=1)
y = df["charges"]
```

## 4. Train-Test Split

Here, we define our features (X) and target (y) and then split the dataset into training and testing sets to evaluate model performance on unseen data. A `random_state` is set for reproducibility.

```python
# Split data into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

## 5. Model Training and Evaluation

We will train various machine learning models and evaluate their performance. Note that 'charges' is a continuous variable, making it primarily a regression problem. Classification models are included for exploration, but their interpretation in this context should consider the `charges` being treated as discrete categories due to `astype(int)` transformation.

### 5.1 Linear Regression Model

Linear Regression is a regression model suitable for predicting continuous numerical values like medical charges. We will evaluate it using Mean Absolute Error (MAE), Mean Squared Error (MSE), and R-squared ($R^2$) score.

# Linear Regression Model

```python
model=LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
prediction = model.predict(x_test)
```

### Evaluating Linear Regression Model Performance

```python
# Calculate evaluation metrics for Linear Regression
mae = mean_absolute_error(y_test, prediction)
mse = mean_squared_error(y_test, prediction)
r2 = r2_score(y_test, prediction)

print(f"Linear Regression - Mean Absolute Error (MAE): {mae:.2f}")
print(f"Linear Regression - Mean Squared Error (MSE): {mse:.2f}")
print(f"Linear Regression - R-squared (R2): {r2:.2f}")
```

### 5.2 Logistic Regression Model

Logistic Regression is typically a classification algorithm. Here, it attempts to classify the integer `charges` values. A `ConvergenceWarning` might indicate that the model struggled to find a stable solution with default parameters, often due to the nature of the target variable (many unique integer classes) or unscaled data.

# Logistic Regression Model

```python
model1 = LogisticRegression(max_iter=1000, random_state=42) # Increased max_iter and added random_state for reproducibility
model1.fit(x_train, y_train)
prediction1 = model1.predict(x_test)
```

```python
model1.fit(x_train, y_train)
```

```python
prediction1 = model1.predict(x_test)
```

### Evaluating Logistic Regression Model Performance with Classification Report

```python
# Calculate classification report for Logistic Regression
report_lr = classification_report(y_test, prediction1)
print("Logistic Regression Classification Report:")
print(report_lr)
```

### 5.3 Decision Tree Classifier Model

Decision Tree Classifier is a non-linear classification model that can capture complex relationships in the data. Its performance will be assessed using a classification report.

# Decision Tree Classifier Model

```python
model2=DecisionTreeClassifier()
```

```python
model2.fit(x_train, y_train)
```

```python
prediction2 = model2.predict(x_test)
```

### Evaluating Decision Tree Classifier Model Performance with Classification Report

```python
# Calculate classification report for Decision Tree Classifier
report_dt = classification_report(y_test, prediction2)
print("Decision Tree Classifier Classification Report:")
print(report_dt)
```

### 5.4 Random Forest Classifier Model

Random Forest is an ensemble learning method that builds multiple decision trees and merges their predictions to improve accuracy and control overfitting. We will evaluate its classification performance.

# Random Forest Classifier Model

```python
model3 = RandomForestClassifier(random_state=42)
model3.fit(x_train, y_train)
prediction3 = model3.predict(x_test)
```

```python
model3.fit(x_train, y_train)
```

```python
prediction3 = model3.predict(x_test)
```

### Evaluating Random Forest Classifier Model Performance with Classification Report

```python
# Calculate classification report for Random Forest Classifier
report_rf = classification_report(y_test, prediction3)
print("Random Forest Classifier Classification Report:")
print(report_rf)
```

### 5.5 K-Means Clustering Model (Unsupervised Learning)

K-Means is an unsupervised clustering algorithm used to group similar data points. Unlike supervised models, it does not use the target variable `y_test` for fitting or direct evaluation metrics like classification reports. Its 'prediction' assigns data points to clusters.

# KMeans Clustering Model

```python
model4 = KMeans(n_clusters=3, random_state=42, n_init='auto')
model4.fit(x_train, y_train)
prediction4 = model4.predict(x_test)
```
