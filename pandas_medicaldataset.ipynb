```python
df=pd.read_csv("content_drive/MyDrive/medical_insurance.csv")
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
coder=LabelEncoder()
```

```python
df["sex"] = df["sex"].astype(str)
df["sex"] = coder.fit_transform(df["sex"])
```

```python
df["smoker"] = df["smoker"].astype(str)
df["smoker"] = coder.fit_transform(df["smoker"])
```

```python
df["region"] = df["region"].astype(str)
df["region"] = coder.fit_transform(df["region"])
```

```python
df["bmi"]=df["bmi"].astype(int)
df["charges"]=df["charges"].astype(int)
```

```python
x=(df.drop("charges",axis=1))
y=df["charges"]
```
